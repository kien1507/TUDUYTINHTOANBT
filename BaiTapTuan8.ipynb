## W7A16
Cho một mảng số nguyên không âm. Gọi
mảng đó là "mảng đẹp" nếu mỗi phần tử
x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp
hay không. Nếu không hãy xóa đi y số (y
là nhỏ nhất có thể) để mảng còn lại thành
mảng đẹp.

```python
def kiem_tra_mang_dep(arr):
    counts = {}
    for x in arr:
        if x in counts:
            counts[x] += 1
        else:
            counts[x] = 1         
    y = 0 
    for x in counts:
        c = counts[x]
        if x == 0:
            y += c    
        elif c < x:
            y += c   
        else:
            y += (c - x)
    if y == 0:
        print("Mảng đã cho là mảng đẹp!")
    else:
        print("Mảng không phải là mảng đẹp.")
        print("Số lượng số cần xóa ít nhất là:", y)
        
    return y
mang_1 = [3, 1, 3, 2, 3, 2]
kiem_tra_mang_dep(mang_1) 
```

## W7A17
Tìm phần tử xuất hiện nhiều hơn n/2 lần
trong mảng (nếu có) với độ phức tạp thời
gian là O(n), độ phức tạp bộ nhớ là O(1),
- Boyer–Moore. Nếu không có phần tử
nào thỏa mãn thì không in gì

```python
def find_majority_element(arr):
    n = len(arr)
    if n == 0:
        return
    candidate = None
    count = 0
    
    for x in arr:
        if count == 0:
            candidate = x
            count = 1
        elif x == candidate:
            count += 1
        else:
            count -= 1
    actual_count = 0
    for x in arr:
        if x == candidate:
            actual_count += 1
    if actual_count > n // 2:
        print(candidate)
    else:
        pass

print("Kết quả test 1: ", end="")
arr1 = [2, 2, 1, 1, 2, 3, 2]
find_majority_element(arr1)
```

## W7A18
Sắp xếp có điều kiện: Số chẵn đứng
trước, số lẻ đứng sau, chẵn tăng dần, lẻ
giảm dần

```python
def sap_xep_co_dieu_kien(arr):
    chan = []
    le = []
    for x in arr:
        if x % 2 == 0:
            chan.append(x)
        else:
            le.append(x)
    chan.sort()               
    le.sort(reverse=True)     
    
    ket_qua = chan + le
    return ket_qua

mang_ban_dau = [5, 2, 9, 1, 6, 4, 8, 3, 7, 0]
print("Mảng ban đầu:", mang_ban_dau)
print("Sau khi sắp xếp:", sap_xep_co_dieu_kien(mang_ban_dau))
```

## W7A19
Merge 2 mảng đã sắp xếp (có thể tăng
dần, có thể giảm dần) mà không dùng
bất cứ thuật toán sắp xếp nào thành một
mảng cũng có cùng thứ tự sắp xếp như 2
mảng ban đầu


```python
def merge(l1,l2):
    i = j = 0
    res = []
    if len(l1) == 0 or len(l2) == 0:
        return l1 + l2
    if l1[0] < l1[-1]:
        while i < len(l1) and j < len(l2):
            if l1[i] < l2[j]:
                res.append(l1[i])
                i += 1
            else:
                res.append(l2[j])
                j += 1
    else:
        while i < len(l1) and j < len(l2):
            if l1[i] > l2[j]: 
                res.append(l1[i])
                i += 1
            else:
                res.append(l2[j])
                j += 1
    while i < len(l1):
        res.append(l1[i])
        i += 1
    while j < len(l2):
        res.append(l2[j])
        j += 1
    return res
l1 = [1, 2, 4, 6, 7, 87]
l2 = [1, 3, 5, 5, 9, 45, 66]

print(*merge(l1, l2))
```

## W7A21
Cho N tên sinh viên gồm cả họ và tên,
tìm xem tên nào hay gặp nhất. Nếu có
nhiều tên cùng số lần gặp thì lấy tên có
thứ tự từ điển nhỏ nhất

```python
n = int(input("Nhập số lượng sinh viên: "))
danh_sach_ten = []
for i in range(n):
    danh_sach_ten.append(input(f"Nhập tên sinh viên thứ {i+1}: "))

# Bước 1: Đếm số lần xuất hiện của từng tên
kho_dem = {}
for ten in danh_sach_ten:
    if ten in kho_dem:
        kho_dem[ten] = kho_dem[ten] + 1
    else:
        kho_dem[ten] = 1

# Bước 2: Tìm tên xuất hiện nhiều nhất và có thứ tự từ điển nhỏ nhất
max_dem = -1
max_ten = ""

for ten, dem in kho_dem.items():
    # Nếu gặp số lần xuất hiện lớn hơn, hoặc bằng nhưng tên nhỏ hơn theo bảng chữ cái
    if dem > max_dem:
        max_dem = dem
        max_ten = ten
    elif dem == max_dem:
        if ten < max_ten:
            max_ten = ten

print("Tên hay gặp nhất là:", max_ten)
```

## W7A22
Đếm số lượng tên trong xâu kí tự. Tên
được định nghĩa là xâu có độ dài >=2 và
có chữ cái đầu viết hoa, các chữ cái khác
viết thường, chỉ có kí tự cái

```python
def dem_ten(xau):
    cac_tu = xau.split()
    dem = 0
    
    for tu in cac_tu:
        if len(tu) >= 2 and tu.isalpha() and tu[0].isupper() and tu[1:].islower():
            dem = dem + 1
            
    return dem

xau_mau = "Nguyen Van A di xe may nhan hieu Honda"
print(dem_ten(xau_mau))
```

## W7A23
Cho xâu kí tự, xóa các từ giống nhau liền
kề, chỉ giữ lại từ đầu tiên

```python
def xoa_tu_lien_ke(s):
    words = s.split()
    if not words:
        return ""
    
    result = words[0]
    
    for i in range(1, len(words)):
        if words[i] != words[i-1]:
            result = result + " " + words[i]
            
    return result

xau_goc = "học học nữa nữa mãi mãi học"
xau_moi = xoa_tu_lien_ke(xau_goc)

print("Xâu gốc:", xau_goc)
print("Xâu sau khi xóa:", xau_moi)
```

## W7A24
Tìm số có đúng k chữ số lớn nhất có thể
trong một số có n chữ số (coi đầu vào là
một số rất lớn - str). Không được hoán
đổi vị trí các chữ số

```python
def tim_so_lon_nhat(s, k):
    n = len(s)
    so_lan_xoa = n - k 
    hop = []
    
    for x in s:
        while len(hop) > 0 and so_lan_xoa > 0 and hop[-1] < x:
            hop.pop()  
            so_lan_xoa = so_lan_xoa - 1
            
        hop.append(x)   
        
    ket_qua = hop[:k]
    return "".join(ket_qua)

print("Số lớn nhất thu được:", tim_so_lon_nhat("14329", 3))
```

## W7A26
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra
xem dãy ngoặc đã cho có hợp lệ hay
không

```python
def kiem_tra_ngoac(chuoi):
    hop = [] 
    for ngoac in chuoi:
        # Nếu gặp dấu mở, bỏ vào hộp
        if ngoac == "(" or ngoac == "[" or ngoac == "{":
            hop.append(ngoac)
        # Nếu gặp dấu đóng, kiểm tra xem dấu mở cuối cùng trong hộp có khớp không
        elif ngoac == ")":
            if len(hop) == 0 or hop[-1] != "(":
                return False
            hop.pop()       
        elif ngoac == "]":
            if len(hop) == 0 or hop[-1] != "[":
                return False
            hop.pop()  
        elif ngoac == "}":
            if len(hop) == 0 or hop[-1] != "{":
                return False
            hop.pop()
            
    return len(hop) == 0

print(kiem_tra_ngoac("([])"))  
print(kiem_tra_ngoac("([)]")) 
```

## W7A27
Bà̀i toán Tìm dãy các hợp số dài nhất liên
tiếp không chứa số nguyên nào <= N (N
là số nguyên dương cho trước). Nếu có
nhiều dãy cùng độ dài thì lấy dãy nhỏ
nhất (gặp đầu tiên từ trái qua)

```python
def la_so_nguyen_to(x):
    if x < 2:
        return False
    for i in range(2, int(x**0.5) + 1):
        if x % i == 0:
            return False
    return True

def tim_day_hop_so(N):
    dai_max = 0
    dau_max = -1
    dai = 0
    dau = -1
    
    for x in range(2, N + 1):
        if la_so_nguyen_to(x) == False:
            if dai == 0:
                dau = x  
            dai = dai + 1
            
        else:
            if dai > dai_max:
                dai_max = dai
                dau_max = dau
            dai = 0  
    if dai > dai_max:
        dai_max = dai
        dau_max = dau
    ket_qua = []
    if dai_max > 0:
        for i in range(dau_max, dau_max + dai_max):
            ket_qua.append(i)
            
    return ket_qua

print(tim_day_hop_so(30)) 
```

## W7A28
Tìm Dãy con tăng dần liên tục dài nhất
của một danh sách nhập vào từ bàn
phím. Lấy dãy đầu tiên nếu có nhiều dãy
có số phần tử bằng nhau

```python
a = [int(x) for x in input("Nhập danh sách: ").split()]

if not a:
    print("Danh sách rỗng")
else:
    dai_max = 1
    dau_max = 0 
    
    dai = 1
    dau = 0     
    
    for i in range(1, len(a)):
        if a[i] > a[i-1]:
            dai = dai + 1
        else:
            if dai > dai_max:
                dai_max = dai
                dau_max = dau
            dai = 1
            dau = i  
    if dai > dai_max:
        dai_max = dai
        dau_max = dau
    ket_qua = a[dau_max : dau_max + dai_max]
    print("Dãy con tăng liên tục dài nhất:", ket_qua)
```

## W7A29
Tìm Dãy con tăng dần KHÔNG liên tục
dài nhất của một danh sách nhập vào từ
bàn phím

```python
a = [int(x) for x in input("Nhập danh sách: ").split()]

if not a:
    print("Danh sách rỗng")
else:
    n = len(a)
    L = [1] * n 
    P = [-1] * n 
    
    for i in range(1, n):
        for j in range(0, i):
            if a[i] > a[j] and L[i] < L[j] + 1:
                L[i] = L[j] + 1
                P[i] = j  
    dai_max = max(L)
    vi_tri_max = L.index(dai_max)
    ket_qua = []
    curr = vi_tri_max
    while curr != -1:
        ket_qua.append(a[curr])
        curr = P[curr]
        
    ket_qua.reverse()
    
    print("Dãy con tăng không liên tục dài nhất:", ket_qua)
```
