## W10A1 [DivideByZero]
Nhập vào hai số nguyên a, b. Thực hiện in ra kết quả phép chia a/b. Bắt lỗi ngoại lệ khi b
bằng 0. Kết quả in ra chính xác tới 2 chữ số thập phân nếu b khác 0. (Dùng try ... except
bắt lỗi ZeroDivisionError)

```python
a = int(input("Nhập số nguyên a: "))
b = int(input("Nhập số nguyên b: "))
try:
    ket_qua = a / b
    print(f"Kết quả phép chia a / b là: {ket_qua:.2f}")
except ZeroDivisionError:
    print("Lỗi: Không thể chia cho số 0!")
```

## W10A2 [Invalid_Index]
Hãy nhập vào một mảng my_list (list) n số nguyên (n nguyên dương). Thực hiện thử lệnh
print(my_list[n]) rồi đọc kỹ Traceback và bắt lỗi Truy cập chỉ số sai. (Dùng try ... except
bắt lỗi IndexError)

```python
try:
    # Nhập số nguyên dương n
    n = int(input("Nhập vào số nguyên dương n: "))
    if n <= 0:
        print("Vui lòng nhập n là một số nguyên dương!")
    else:
        my_list = []
        for i in range(n):
            val = int(input(f"Nhập phần tử thứ {i}: "))
            my_list.append(val)
        
        print("\nMảng vừa nhập là:", my_list)
        print(f"Cố gắng truy cập phần tử my_list[{n}]...")
        print(my_list[n])

except IndexError as e:
    print("\n--- PHÁT HIỆN LỖI (Traceback) ---")
    print(f"Lỗi: {e}")
    print("Giải thích: Bạn đã truy cập vào chỉ số vượt quá phạm vi của mảng (Invalid Index).")
except ValueError:
    print("Lỗi: Vui lòng chỉ nhập số nguyên hợp lệ!")
```
