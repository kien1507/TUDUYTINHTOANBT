## W8A17
Yêu cầu
Xây dựng lớp Vehicle(brand, speed).
Xây dựng lớp Car(brand, speed, seats) kế thừa từ Vehicle.
Viết phương thức info() cho mỗi lớp:
- Vehicle: brand-speed
- Car: brand-speed-seats
Input
Dòng 1: loại đối tượng (Vehicle hoặc Car)
Dòng 2: dữ liệu tương ứng.
Output
In ra chuỗi info()

```python
class Vehicle:
    def __init__(self, brand, speed):
        self.b = brand
        self.sp = speed
    def info(self):
        return f"{self.b} - {self.sp}"
class Car(Vehicle):
    def __init__(self, brand, speed, seats):
        super().__init__(brand, speed)
        self.se = seats
    def info(self):
        return f"{self.b} - {self.sp} - {self.se}"

brand, speed, seats = map(str, input().split())
Xe = Vehicle(brand, speed)
Ôtô = Car(brand, speed, seats)
print(Xe.info())
print(Ôtô.info())
    
```

## W8A18
Yêu cầu
Xây dựng lớp Polygon với phương thức perimeter().
Xây dựng lớp Triangle(a, b, c) kế thừa từ Polygon và ghi đè perimeter() trả về a+b+c.
Input
Một dòng chứa 3 số nguyên a b c.
Output
In ra chu vi tam giác. 

```python
class Polygon:
    def perimeter(self):
        return 0

class Triangle(Polygon):
    def __init__(self, aa, bb, cc):
        self.a = aa
        self.b = bb
        self.c = cc
    def perimeter(self):   # Ghi đè phương thức
        return self.a + self.b + self.c

a, b, c = map(int, input().split())
tam_giac = Triangle(a, b, c)
print(tam_giac.perimeter())
```

## W8A22
Yêu cầu
Xây dựng lớp Matrix2x2(a, b, c, d) biểu diễn ma trận:
[a b]
[c d]
Viết phương thức trace() trả về a + d và determinant() trả về ad - bc.
Input
Một dòng chứa 4 số nguyên a b c d.
Output
In ra trace và determinant trên cùng một dòng. 

```python
class Matrix2x2:
    def __init__(self,aa,bb,cc,dd):
        self.a, self.b, self.c, self.d = aa, bb, cc, dd
    def trace(self):
        return self.a + self.d
    def determinant(self):
        return self.a*self.d - self.b*self.c

a, b, c, d = map(int, input().split())

m = Matrix2x2(a, b, c, d)
print(m.trace(), m.determinant())
```

Input

Output

## W8A24
Yêu cầu
Xây dựng lớp Polynomial2 biểu diễn đa thức bậc hai ax^2 + bx + c.
Viết phương thức evaluate(x) để tính giá trị đa thức tại x.
Input
a b c x
Output
Giá trị của đa thức. 

```python
class Polynomial2:
    def __init__(self,aa,bb,cc):
        self.a, self.b, self.c = aa, bb, cc
    def evaluate(self, x):
        return self.a * (x ** 2) + self.b * x + self.c

a, b, c, x = map(float, input().split())

da_thuc = Polynomial2(a, b, c)
print(da_thuc.evaluate(x))
```

## W8A25
Yêu cầu
Xây dựng lớp PolynomialSum cho hai đa thức bậc nhất:
P(x) = ax + b, Q(x) = cx + d.
Viết phương thức sum_str() trả về chuỗi biểu diễn đa thức tổng dưới dạng kx + m.
Input
a b c d
Output
Chuỗi của đa thức P(x)+Q(x). 

```python
class PolynomialSum:
    def __init__(self,aa,bb):
        self.a = aa
        self.b = bb
    def __str__(self):
        return f"{self.a}x + {self.b}"
    def sum_str(self,func):
        k = self.a + func.a
        m = self.b + func.b
        return f"{k}x + {m}"
    
a, b, c, d = map(int, input().split())
p = PolynomialSum(a, b)
q = PolynomialSum(c, d)
print(p)
print(q)
print(p.sum_str(q))
```

## W8A29
Yêu cầu
Xây dựng lớp Time(h, m, s) với phương thức to_seconds() đổi thời gian sang tổng số giây.
Input
h m s
Output
Tổng số giây

```python
class Time:
    def __init__(self,hh,mm,ss):
        self.h, self.m, self.s = hh, mm, ss
    def to_seconds(self):
        return self.h*3600 + self.m*60 + self.s
h, m, s = map(int, input().split())
time1 = Time(h, m, s)
print(time1.to_seconds())
```
