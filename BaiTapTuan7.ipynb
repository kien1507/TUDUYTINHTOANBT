## W7A1
Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần.
Trả về chỉ số của phần tử hoặc -1 nếu không tìm thấy. mẫu hàm: def
binary_search(arr, left, right, target):

```python
def binary_search(arr,left,right, target):
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid -1
    return -1
arr = list(map(int, input().split()))
binary_search(arr,0,len(arr) - 1,3)
```

## W7A3
Viết hàm thực hiện thuật toán bubble sort, Selection Sort để sắp xếp danh sách theo thứ tự tăng dần. 
Sắp xếp nổi bọt (Bubble Sort); Sắp xếp chọn (SelectionSort)

```python
def bubble_sort(lst):
    n = len(lst)
    for i in range(n):
        swapped = False
        for j in range(0,n - i -1):
            if lst[j] > lst[j+1]:
                lst[j], lst[j+1] = lst[j+1], lst[j]
                swapped = True
        if not swapped:
            break
    return lst


def selection_sort(lst):
    n = len(lst)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if lst[j] > lst[min_idx]:
                min_idx = j
        lst[i], lst[min_idx] = lst[min_idx], lst[i]
    return lst
```

Cho danh sách các số nguyên và một số X, đếm
tất cả các cặp (a, b) sao cho a + b = X. c

```python
def count_pairs_brute_force(arr, X):
    count = 0
    n = len(arr)
    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] + arr[j] == X:
                count += 1
    return count
```

## W7A9
Cho một danh sách các số nums, với mỗi phần tử
nums[i], hãy đếm xem có bao nhiêu phần tử bên phải
nhỏ hơn nums[i].
Trả về danh sách các số đếm.
Efficient solution: using heap or binary search tree

```python
def lst_so_dem(lst):
    n = len(lst)
    res = []
    for i in range(n):
        c = 0
        for j in range(i+1, n):
            if lst[j] < lst[i]:
                c += 1
        res.append(c)
    return res
lst = [3,4,6,2,7,1,8]
lst_so_dem(lst)
```

## W7A10
Cho một danh sách các số nguyên nums và một số
nguyên m, hãy tìm tổng mảng con lớn nhất theo
modulo m.
Tức là, với bất kỳ mảng con nums[i:j] nào, hãy tính
sum(nums[i:j]) % m và trả về kết quả lớn nhất có thể.
Gợi ý:
Duy trì tổng tiền tố (prefix sum) mod m và sử dụng
tìm kiếm nhị phân để tìm tổng tiền tố nhỏ nhất chỉ
lớn hơn tổng hiện tại một chút (để tối đa hóa hiệu số
bao quanh).


```python
def giai_bai_toan(nums, m):
    max_modulo = 0
    n = len(nums)
    for i in range(n):
        tong_hien_tai = 0
        for j in range(i, n):
            tong_hien_tai = (tong_hien_tai + nums[j]) % m
            if tong_hien_tai > max_modulo:
                max_modulo = tong_hien_tai
    return max_modulo
nums = [3, 3, 9, 9, 5]
m = 7
print(giai_bai_toan(nums, m)) 
```

## W7A13
Dãy hài hòa là một dãy thỏa mãn điều kiện giá trị lớn nhất và giá trị nhỏ nhất của dãy khác nhau đúng
bằng \( 1 \). Viết hàm trả về độ dài của dãy hài hòa con dài nhất có thể có trong mảng giá trị nguyên \( arr \) có \( n \) phần tử
Ví dụ:
Input: 1 3 2 2 5 2 3 7
Output: 5 - Dãy hài hòa con
dài nhất là [3,2,2,2,3]


```python
def find_longest_harmonious_sequence(n, arr):
    counts = {}
    for x in arr:
        if x in counts:
            counts[x] += 1
        else:
            counts[x] = 1
    
    max_len = 0
    for x in counts:
        if (x + 1) in counts:
            current_len = counts[x] + counts[x + 1]
            if current_len > max_len:
                max_len = current_len
                
    return max_len

n = int(input())
arr = list(map(int, input().split()))
find_longest_harmonious_sequence(n, arr)
    
```

Nemo được mời tới tham dự hội nghị bàn tròn giữa các nhân vật hoạt hình. Tất cả các khách mời sẽ được gửi thiệp mời kèm theo một danh sách khách mời. Nemo muốn biết các nhân vật ngồi cạnh mình là ai để có thể chuẩn bị quà cho họ. Hãy giúp Nemo tìm tên của các nhân vật này, biết rằng tên trong danh sách khách mời được sắp xếp theo vị trí ngồi.
  Đầu vào: số nguyên n - số lượng khách mời trong danh sách trên một dòng, n dòng tiếp theo là tên của các vị khách mời.
  Đầu ra: Tên của hai người ngồi cạnh Nemo theo định dạng "<name1> and <name2>" (name1 là tên của người đứng trước và name2 là tên của người đứng sau Nemo trong danh sách). 


```python

n = int(input())    # Nhập số lượng khách mời

danh_sach = []      # Nhập danh sách tên các khách mời
for i in range(n):
    ten = input().strip()
    danh_sach.append(ten)

idx = danh_sach.index("Nemo")       # Tìm vị trí (chỉ số index) của Nemo trong danh sách

nguoi_truoc = danh_sach[(idx - 1) % n]      # Tìm người đứng trước và đứng sau dựa vào tính chất bàn tròn (% n)
                                            # Ví dụ: Nếu idx = 0 thì (0 - 1) % n sẽ trả về n - 1 (người cuối cùng)

nguoi_sau = danh_sach[(idx + 1) % n]    # Nếu idx = n - 1 thì (idx + 1) % n sẽ trả về 0 (người đầu tiên)

print(f"{nguoi_truoc} and {nguoi_sau}")     # In kết quả theo đúng định dạng đề bài yêu cầu
```
